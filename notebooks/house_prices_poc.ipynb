charger les bibliothèques

```python

import pickle
import numpy as np
import pandas as pd
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

print("pandas :", pd.__version__)
print("scikit-learn :", sklearn.__version__)
```

charger train.csv

```python

df = pd.read_csv("/kaggle/input/competitions/house-prices-advanced-regression-techniques/train.csv")
print(df.shape)
df.head()
```

NETT:garder les variables numériques et supprimer les lignes incomplètes

```python

nb_avant = len(df)

df_num = df.select_dtypes(include=[np.number])
print("Nombre de colonnes numériques :", df_num.shape[1])

df_num = df_num.dropna()
print("Lignes avant :", nb_avant)
print("Lignes après :", len(df_num))
print("Lignes supprimées :", nb_avant - len(df_num))
```

séparer X et y

```python

y = df_num["SalePrice"]
X = df_num.drop(columns=["SalePrice", "Id"])  # Id est un identifiant, pas une caractéristique du bien

print("X :", X.shape)
print("y :", y.shape)
```

découper en train et test avec stratification par binning de la cible

```python

bins = pd.qcut(y, q=10, labels=False)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=bins,
    random_state=42
)

print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
print("Prix moyen train :", round(y_train.mean()))
print("Prix moyen test  :", round(y_test.mean()))
```

entraîner la régression linéaire

```python

model = LinearRegression()
model.fit(X_train, y_train)

print("Modèle entraîné")
print("Nombre de coefficients :", len(model.coef_))
```

mesurer RMSE, MAE et R² sur train et sur test

```python

y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

rmse_train = np.sqrt(mean_squared_error(y_train, y_pred_train))
rmse_test = np.sqrt(mean_squared_error(y_test, y_pred_test))
mae_train = mean_absolute_error(y_train, y_pred_train)
mae_test = mean_absolute_error(y_test, y_pred_test)
r2_train = r2_score(y_train, y_pred_train)
r2_test = r2_score(y_test, y_pred_test)

print(f"TRAIN : RMSE = {rmse_train:,.0f} $ | MAE = {mae_train:,.0f} $ | R² = {r2_train:.4f}")
print(f"TEST  : RMSE = {rmse_test:,.0f} $ | MAE = {mae_test:,.0f} $ | R² = {r2_test:.4f}")
```

sauvegarder le modèle dans model.pkl

```python

with open("model.pkl", "wb") as f:
    pickle.dump(model, f)

print("model.pkl sauvegardé")
```

```python

print("=== À INSCRIRE AU TABLEAU ===")
print(f"RMSE test : {rmse_test:,.0f} $")
print(f"MAE test  : {mae_test:,.0f} $")
print(f"R² test   : {r2_test:.4f}")
print()
print("pandas       :", pd.__version__)
print("scikit-learn :", sklearn.__version__)
print()
print("=== POUR LA DISCUSSION ===")
print("Lignes avant dropna :", len(df))
print("Lignes après dropna :", len(df_num))
print("Nombre de bins      : 10")
print("random_state        : 42")
print("test_size           : 0.2")
```
